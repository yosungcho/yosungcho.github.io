```python
import os
import pandas as pd
import numpy as np
from PIL import Image
from pathlib import Path
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import keras
from keras.layers import Dense
# Function to load and process the images
def load_and_process_images(folder_path, df):
    images = []
    labels = []
    # Ensure the 'imageID' column in the DataFrame is in lowercase for consistent matching
    df['imageID_lower'] = df['imageID'].str.strip().str.lower()
    for image_file in os.listdir(folder_path):
        if image_file.endswith('.jpg'):
            image_id = image_file.strip().lower() # Remove leading/trailing whitespaces and convert to lowercase
            try:
                # Match image_id with the lowercase 'imageID_lower' column
                label = df.loc[df['imageID_lower'] == image_id, 'binaryLabels'].values[0]
                # Load the image using PIL
                image_path = Path(folder_path) / image_file
                image = Image.open(image_path)
                image = image.resize((128, 128))
                image = np.array(image) / 255.0 # Resize the image to a fixed size (you can adjust this size as needed)
                # Normalize the pixel values to the range [0, 1]
                images.append(image)
                labels.append(label)
            except IndexError:
                print(f"Error: No matching label found for image {image_id}")
                continue # Skip to the next image if no label is found

    images = np.array(images)
    labels = np.array(labels)
    return images, labels
# Replace 'folder_path' with the path to your image folder
folder_path = '/content/drive/MyDrive/Jojo CNN/CNN G1020 Run/962 images'
# Load the data from the Excel file into a DataFrame
label_file = '/content/drive/MyDrive/Jojo CNN/CNN G1020 Run/962 images.xlsx'
df = pd.read_excel(label_file)
# Load and process the images
images, labels = load_and_process_images(folder_path, df)
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(images, labels, test_size=0.2, random_state=42)
# Create the CNN model
model = Sequential()
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(128, 128, 3)))
model.add(MaxPooling2D((2, 2)))
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D((2, 2)))
model.add(Flatten())
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
# Step 2: Compile the model
optimizer = keras.optimizers.Adam(learning_rate=0.0002) # Adjust the learning rate if needed
model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy'])
history = model.fit(X_train, y_train, batch_size=32, epochs=25, validation_split=0.2)

# Step 5: Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test)
print("Test loss:", loss)
print("Test accuracy:", accuracy)
# Assuming you have loaded your data and split it into training and validation sets
# Get the number of images and labels for training
num_train_images = len(X_train)
num_train_labels = len(y_train)
# Get the number of images and labels for validation
num_val_images = len(X_test)
num_val_labels = len(y_test)
# Print the results
print("Number of images for training:", num_train_images)
print("Number of labels for training:", num_train_labels)
print("Number of images for validation:", num_val_images)
print("Number of labels for validation:", num_val_labels)
```
