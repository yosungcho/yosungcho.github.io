```python
"""
============================================================================
AMD: Regenerate Predictions + Threshold Analysis (Standalone)
============================================================================

Runs from a fresh Colab session with no prior state needed.
Loads the trained model checkpoint, regenerates test predictions,
then runs comprehensive threshold analysis with data-driven decision points.

Requires only:
  - Drive mounted
  - best_model.pth in /content/drive/MyDrive/FundusView/amd/checkpoints/
  - ODIR-5K images in /content/drive/MyDrive/FundusView/ODIR-5K/preprocessed_images/
  - A split_test_*.csv file in results/  (from the original training run)
============================================================================
"""

# ============================================================================
# SETUP
# ============================================================================

import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms
from sklearn.metrics import (
    roc_curve, precision_recall_curve, confusion_matrix,
    roc_auc_score, average_precision_score
)
from tqdm import tqdm

# Mount Drive
from google.colab import drive
try:
    os.listdir('/content/drive/MyDrive')
    print("Drive already mounted")
except:
    drive.mount('/content/drive')

# Install/import timm
try:
    import timm
except ImportError:
    os.system('pip install timm --quiet')
    import timm

# Paths
DRIVE_ROOT = '/content/drive/MyDrive/FundusView'
PROJECT_DIR = os.path.join(DRIVE_ROOT, 'amd')
CHECKPOINT_DIR = os.path.join(PROJECT_DIR, 'checkpoints')
RESULTS_DIR = os.path.join(PROJECT_DIR, 'results')
IMAGE_DIR = os.path.join(DRIVE_ROOT, 'ODIR-5K', 'preprocessed_images')

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")


# ============================================================================
# LOAD TEST SPLIT
# ============================================================================

test_splits = sorted(glob.glob(os.path.join(RESULTS_DIR, 'split_test_*.csv')))
if not test_splits:
    raise FileNotFoundError(
        f"No test split CSV found in {RESULTS_DIR}. "
        "The original training script should have saved these. "
        "If not, you need to re-run at least Section 3 of the training script "
        "(the data prep step) to regenerate the test split."
    )

test_split_path = test_splits[-1]
df_test = pd.read_csv(test_split_path)
RUN_ID = os.path.basename(test_split_path).replace('split_test_', '').replace('.csv', '')

print(f"Loaded test split: {os.path.basename(test_split_path)}")
print(f"Test set: {len(df_test)} images, {int(df_test['amd'].sum())} AMD-positive")


# ============================================================================
# REBUILD MODEL AND LOAD CHECKPOINT
# ============================================================================

class AMDBinaryModel(nn.Module):
    def __init__(self, backbone_name='efficientnet_b2', pretrained=False, dropout=0.3):
        super().__init__()
        self.backbone = timm.create_model(
            backbone_name, pretrained=pretrained,
            num_classes=0, global_pool='avg'
        )
        self.amd_head = nn.Sequential(
            nn.Linear(self.backbone.num_features, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),
            nn.Linear(256, 1)
        )
    def forward(self, x):
        return self.amd_head(self.backbone(x)).squeeze(-1)


checkpoint_path = os.path.join(CHECKPOINT_DIR, 'best_model.pth')
if not os.path.exists(checkpoint_path):
    raise FileNotFoundError(f"No checkpoint at {checkpoint_path}")

print(f"\nLoading model from {checkpoint_path}...")
model = AMDBinaryModel().to(device)
best_ckpt = torch.load(checkpoint_path, map_location=device, weights_only=False)
model.load_state_dict(best_ckpt['model_state_dict'])
model.eval()
print(f"Loaded best model from epoch {best_ckpt['epoch']} with val AUROC {best_ckpt['val_auroc']:.4f}")


# ============================================================================
# RUN INFERENCE ON TEST SET
# ============================================================================

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

eval_transform = transforms.Compose([
    transforms.Resize((512, 512)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])


class ODIRAMDDataset(Dataset):
    def __init__(self, df, image_dir, transform=None):
        self.df = df.reset_index(drop=True)
        self.image_dir = image_dir
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        image = Image.open(os.path.join(self.image_dir, str(row['filename']))).convert('RGB')
        if self.transform:
            image = self.transform(image)
        return image, torch.tensor(row['amd'], dtype=torch.float32)


test_dataset = ODIRAMDDataset(df_test, IMAGE_DIR, transform=eval_transform)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=2, pin_memory=True)

print("\nRunning inference on test set (this takes ~3 min)...")
all_probs, all_labels = [], []
with torch.no_grad():
    for images, labels in tqdm(test_loader):
        images = images.to(device, non_blocking=True)
        logits = model(images)
        probs = torch.sigmoid(logits).cpu().numpy()
        all_probs.extend(probs.tolist())
        all_labels.extend(labels.numpy().tolist())

test_probs = np.array(all_probs)
test_labels = np.array(all_labels)

# Save predictions immediately so we don't have to regenerate again
predictions_path = os.path.join(RESULTS_DIR, f'test_predictions_{RUN_ID}.csv')
df_out = df_test.copy()
df_out['amd_probability'] = test_probs
df_out.to_csv(predictions_path, index=False)
print(f"\nPredictions saved to: {predictions_path}")


# ============================================================================
# THRESHOLD ANALYSIS
# ============================================================================

print("\n" + "=" * 60)
print("PREDICTION DISTRIBUTION ANALYSIS")
print("=" * 60)

n_total = len(test_probs)
n_amd = int(test_labels.sum())
n_normal = n_total - n_amd
amd_probs = test_probs[test_labels == 1]
normal_probs = test_probs[test_labels == 0]

print(f"Total: {n_total} images | AMD: {n_amd} ({n_amd/n_total*100:.1f}%) | Non-AMD: {n_normal}")

print(f"\nAMD prediction stats:")
print(f"  Min:    {amd_probs.min():.4f}")
print(f"  25%:    {np.percentile(amd_probs, 25):.4f}")
print(f"  Median: {np.median(amd_probs):.4f}")
print(f"  75%:    {np.percentile(amd_probs, 75):.4f}")
print(f"  Max:    {amd_probs.max():.4f}")
print(f"  Fraction < 0.1: {(amd_probs < 0.1).sum() / len(amd_probs) * 100:.1f}%  ← 'invisible AMD'")
print(f"  Fraction > 0.5: {(amd_probs > 0.5).sum() / len(amd_probs) * 100:.1f}%  ← 'obvious AMD'")

print(f"\nNon-AMD prediction stats:")
print(f"  Median: {np.median(normal_probs):.4f}")
print(f"  95%ile: {np.percentile(normal_probs, 95):.4f}")
print(f"  99%ile: {np.percentile(normal_probs, 99):.4f}")


# ============================================================================
# CANDIDATE THRESHOLDS
# ============================================================================

fpr, tpr, roc_thresholds = roc_curve(test_labels, test_probs)
precision, recall, pr_thresholds = precision_recall_curve(test_labels, test_probs)

J = tpr - fpr
thr_youden = roc_thresholds[np.argmax(J)]

thr_99spec = roc_thresholds[np.argmin(np.abs((1 - fpr) - 0.99))]
thr_95spec = roc_thresholds[np.argmin(np.abs((1 - fpr) - 0.95))]
thr_90spec = roc_thresholds[np.argmin(np.abs((1 - fpr) - 0.90))]
thr_90sens = roc_thresholds[np.argmin(np.abs(tpr - 0.90))]
thr_80sens = roc_thresholds[np.argmin(np.abs(tpr - 0.80))]

f1_scores = 2 * precision[:-1] * recall[:-1] / (precision[:-1] + recall[:-1] + 1e-8)
thr_f1 = pr_thresholds[np.argmax(f1_scores)] if len(pr_thresholds) > 0 else 0.5

candidates = [
    ("99% specificity", thr_99spec),
    ("95% specificity", thr_95spec),
    ("Youden's J max", thr_youden),
    ("90% specificity", thr_90spec),
    ("F1-optimal", thr_f1),
    ("80% sensitivity", thr_80sens),
    ("90% sensitivity", thr_90sens),
    ("Default (0.5)", 0.5),
]

print("\n" + "=" * 100)
print(f"{'Operating point':<20} {'Threshold':<12} {'Sens':<8} {'Spec':<8} {'PPV':<8} {'Catches AMD':<18} {'False alarms'}")
print("-" * 100)

for name, thr in candidates:
    preds = (test_probs >= thr).astype(int)
    cm = confusion_matrix(test_labels.astype(int), preds)
    tn, fp, fn, tp = cm.ravel()
    sens = tp / (tp + fn) if (tp + fn) > 0 else 0
    spec = tn / (tn + fp) if (tn + fp) > 0 else 0
    ppv = tp / (tp + fp) if (tp + fp) > 0 else 0
    print(f"{name:<20} {thr:<12.4f} {sens:<8.3f} {spec:<8.3f} {ppv:<8.3f} "
          f"{f'{tp}/{tp+fn} = {sens*100:.0f}%':<18} {f'{fp} of {tn+fp}'}")


# ============================================================================
# VISUALIZATION (6 plots)
# ============================================================================

fig = plt.figure(figsize=(18, 12))

# --- Plot 1: Strip plot ---
ax1 = plt.subplot(2, 3, 1)
np.random.seed(42)
jitter_normal = np.random.uniform(-0.15, 0.15, size=len(normal_probs))
jitter_amd = np.random.uniform(-0.15, 0.15, size=len(amd_probs))

ax1.scatter(normal_probs, np.zeros(len(normal_probs)) + jitter_normal,
            s=4, alpha=0.15, c='steelblue', label=f'Non-AMD (n={len(normal_probs)})')
ax1.scatter(amd_probs, np.ones(len(amd_probs)) + jitter_amd,
            s=20, alpha=0.6, c='crimson', label=f'AMD (n={len(amd_probs)})')

invisible_amd = amd_probs[amd_probs < 0.1]
ax1.scatter(invisible_amd, np.ones(len(invisible_amd)) + np.random.uniform(-0.15, 0.15, len(invisible_amd)),
            s=40, facecolors='none', edgecolors='darkred', linewidths=1.5,
            label=f'Missed AMD (prob<0.1): {len(invisible_amd)}')

ax1.set_xlabel('Predicted AMD Probability')
ax1.set_yticks([0, 1])
ax1.set_yticklabels(['Non-AMD', 'AMD'])
ax1.set_title('Every Test Prediction (strip plot)')
ax1.legend(loc='center right', fontsize=9)
ax1.grid(True, alpha=0.3, axis='x')
ax1.set_xlim(-0.02, 1.02)

# --- Plot 2: Log-scale distribution ---
ax2 = plt.subplot(2, 3, 2)
bins = np.logspace(-5, 0, 50)
normal_clipped = np.clip(normal_probs, 1e-5, 1.0)
amd_clipped = np.clip(amd_probs, 1e-5, 1.0)
ax2.hist(normal_clipped, bins=bins, alpha=0.5, color='steelblue', label='Non-AMD', density=True)
ax2.hist(amd_clipped, bins=bins, alpha=0.5, color='crimson', label='AMD', density=True)
ax2.set_xscale('log')
ax2.set_xlabel('Predicted AMD Probability (log scale)')
ax2.set_ylabel('Density')
ax2.set_title('Distribution on Log Scale\n(exposes AMD clumped near 0)')
ax2.legend()
ax2.grid(True, alpha=0.3)

# --- Plot 3: Cumulative distribution ---
ax3 = plt.subplot(2, 3, 3)
thresholds_range = np.linspace(0, 1, 500)
frac_normal_below = [(normal_probs < t).sum() / len(normal_probs) for t in thresholds_range]
frac_amd_below = [(amd_probs < t).sum() / len(amd_probs) for t in thresholds_range]

ax3.plot(thresholds_range, frac_normal_below, color='steelblue', linewidth=2,
         label='Non-AMD (want high = specificity)')
ax3.plot(thresholds_range, frac_amd_below, color='crimson', linewidth=2,
         label='AMD (want low = sensitivity)')
ax3.fill_between(thresholds_range, frac_normal_below, frac_amd_below,
                 where=[n > a for n, a in zip(frac_normal_below, frac_amd_below)],
                 alpha=0.2, color='green', label='Separation area')
ax3.set_xlabel('Threshold')
ax3.set_ylabel('Fraction of class below threshold')
ax3.set_title('Cumulative Distribution\n(area between curves = separation)')
ax3.legend(loc='center right', fontsize=9)
ax3.grid(True, alpha=0.3)

# --- Plot 4: Metrics vs threshold ---
ax4 = plt.subplot(2, 3, 4)
sens_curve, spec_curve, ppv_curve = [], [], []
for t in thresholds_range:
    preds = (test_probs >= t).astype(int)
    tp = ((preds == 1) & (test_labels == 1)).sum()
    tn = ((preds == 0) & (test_labels == 0)).sum()
    fp = ((preds == 1) & (test_labels == 0)).sum()
    fn = ((preds == 0) & (test_labels == 1)).sum()
    sens_curve.append(tp / (tp + fn) if (tp + fn) > 0 else 0)
    spec_curve.append(tn / (tn + fp) if (tn + fp) > 0 else 0)
    ppv_curve.append(tp / (tp + fp) if (tp + fp) > 0 else 0)

ax4.plot(thresholds_range, sens_curve, color='crimson', linewidth=2, label='Sensitivity')
ax4.plot(thresholds_range, spec_curve, color='steelblue', linewidth=2, label='Specificity')
ax4.plot(thresholds_range, ppv_curve, color='darkgreen', linewidth=2, label='PPV')

for name, thr in [('95% spec', thr_95spec), ("Youden", thr_youden), ('80% sens', thr_80sens)]:
    ax4.axvline(thr, linestyle='--', alpha=0.5, color='gray')
    ax4.text(thr, 0.02, f' {name}\n {thr:.3f}', rotation=90, fontsize=8, va='bottom')

ax4.set_xlabel('Threshold')
ax4.set_ylabel('Metric Value')
ax4.set_title('Metrics vs Threshold')
ax4.legend(loc='center right', fontsize=9)
ax4.grid(True, alpha=0.3)
ax4.set_ylim(-0.02, 1.02)

# --- Plot 5: ROC with candidates ---
ax5 = plt.subplot(2, 3, 5)
ax5.plot(fpr, tpr, color='navy', linewidth=2,
         label=f'ROC (AUROC={roc_auc_score(test_labels, test_probs):.3f})')
ax5.plot([0, 1], [0, 1], 'k--', alpha=0.3)

colors = plt.cm.tab10(np.linspace(0, 1, len(candidates)))
for (name, thr), color in zip(candidates, colors):
    preds = (test_probs >= thr).astype(int)
    tp = ((preds == 1) & (test_labels == 1)).sum()
    tn = ((preds == 0) & (test_labels == 0)).sum()
    fp = ((preds == 1) & (test_labels == 0)).sum()
    fn = ((preds == 0) & (test_labels == 1)).sum()
    spec = tn / (tn + fp) if (tn + fp) > 0 else 0
    sens = tp / (tp + fn) if (tp + fn) > 0 else 0
    ax5.scatter(1 - spec, sens, s=120, color=color, zorder=5,
                edgecolors='black', linewidths=1, label=f'{name} (t={thr:.3f})')

ax5.set_xlabel('False Positive Rate')
ax5.set_ylabel('True Positive Rate')
ax5.set_title('ROC with All Candidate Points')
ax5.legend(loc='lower right', fontsize=8)
ax5.grid(True, alpha=0.3)

# --- Plot 6: Tier distributions ---
ax6 = plt.subplot(2, 3, 6)
tier_configs = [
    ("Default\n(0.30/0.60)", 0.30, 0.60),
    ("Aggressive\n(0.05/0.30)", 0.05, 0.30),
    ("Conservative\n(0.10/0.50)", 0.10, 0.50),
    ("Data-driven\n(P95_norm/thr_80sens)",
     np.percentile(normal_probs, 95), thr_80sens),
    ("Screening-opt\n(thr_90sens/thr_95spec)",
     min(thr_90sens, thr_95spec), max(thr_90sens, thr_95spec)),
]

x_positions = np.arange(len(tier_configs))
normal_pcts, uncertain_pcts, amd_pcts = [], [], []

for label, low_t, high_t in tier_configs:
    tiers = np.where(test_probs < low_t, 'Normal',
            np.where(test_probs >= high_t, 'AMD', 'Uncertain'))
    normal_pcts.append((tiers == 'Normal').sum() / len(tiers) * 100)
    uncertain_pcts.append((tiers == 'Uncertain').sum() / len(tiers) * 100)
    amd_pcts.append((tiers == 'AMD').sum() / len(tiers) * 100)

ax6.bar(x_positions, normal_pcts, color='steelblue', label='Normal')
ax6.bar(x_positions, uncertain_pcts, bottom=normal_pcts, color='orange', label='Uncertain')
ax6.bar(x_positions, amd_pcts, bottom=[n+u for n, u in zip(normal_pcts, uncertain_pcts)],
        color='crimson', label='AMD Alert')

ax6.set_xticks(x_positions)
ax6.set_xticklabels([c[0] for c in tier_configs], fontsize=8)
ax6.set_ylabel('% of test predictions')
ax6.set_title('Three-Tier Distribution\nfor different threshold pairs')
ax6.legend(loc='upper right', fontsize=8)
ax6.grid(True, alpha=0.3, axis='y')

print("\n" + "=" * 60)
print("THREE-TIER WARNING DISTRIBUTIONS")
print("=" * 60)
for (label, low_t, high_t), n_pct, u_pct, a_pct in zip(tier_configs, normal_pcts, uncertain_pcts, amd_pcts):
    label_clean = label.replace('\n', ' ')
    tiers = np.where(test_probs < low_t, 'Normal',
            np.where(test_probs >= high_t, 'AMD', 'Uncertain'))
    amd_in_normal = ((tiers == 'Normal') & (test_labels == 1)).sum()
    amd_in_uncertain = ((tiers == 'Uncertain') & (test_labels == 1)).sum()
    amd_in_alert = ((tiers == 'AMD') & (test_labels == 1)).sum()
    total_flagged = amd_in_alert + amd_in_uncertain

    print(f"\n{label_clean}: lower={low_t:.4f}, upper={high_t:.4f}")
    print(f"  Distribution: Normal {n_pct:.1f}% | Uncertain {u_pct:.1f}% | Alert {a_pct:.1f}%")
    print(f"  AMD cases: Alert={amd_in_alert}, Uncertain={amd_in_uncertain}, Missed={amd_in_normal}")
    print(f"  → Total AMD flagged: {total_flagged}/{n_amd} ({total_flagged/n_amd*100:.1f}%)")

plt.tight_layout()
plot_path = os.path.join(RESULTS_DIR, 'threshold_analysis.png')
plt.savefig(plot_path, dpi=150, bbox_inches='tight')
plt.show()
print(f"\nAnalysis plot saved: {plot_path}")
print(f"Predictions saved: {predictions_path}")

print("\n" + "=" * 60)
print("KEY QUESTIONS TO ASK YOURSELF")
print("=" * 60)
print("""
1. Looking at the log-scale histogram (top-middle):
   - Are AMD cases exactly at 0.0, or clustered around 0.01-0.05?
   - If clustered slightly above 0, aggressive threshold can rescue them
   - If exactly at 0, only more training data will help

2. Looking at the strip plot (top-left):
   - How many red dots at far left (missed AMD)?
   - Are the AMD probabilities generally shifted right of normal?

3. Looking at tier distributions (bottom-right):
   - Which threshold pair flags the most AMD cases into
     Alert+Uncertain (i.e., recommends follow-up)?
   - What percentage of non-AMD patients would that config
     over-refer (Uncertain rate)?

Post the log-scale histogram observation back and I'll recommend
the specific threshold pair for your deployment.
""")
```
